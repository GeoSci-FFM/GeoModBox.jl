# Thermal Convection

Now that we can solve the two-dimensional conservation equations of momentum, energy, and mass using the finite-difference method, we can turn our attention to the problem of thermal convection. Thermal convection provides an excellent example for studying heat transport through the combined effects of advection and diffusion. We begin with the simplest form of thermal convection: an isoviscous, bottom-heated convection model. For the time being, the governing equations are considered in their dimensional form.

# The Problem

Thermal convection is primarily driven by density variations caused by temperature differences and the resulting buoyancy forces. When material is heated from **below** (or, in some cases, also by **internal heat production**), it expands, becomes less dense, and begins to rise. Near the top of the domain, heat is transferred to the surface by thermal diffusion, causing the material to cool, contract, and increase in density. As a result, the colder and denser material sinks. Together, these processes form the fundamental driving mechanism of thermal convection (Fig. 1).

<img src="./Figures/Exercise11_1.png" alt="Sketch of thermal convection" width="450"/> <br>
**Figure 1.** Schematic illustration of a bottom-heated thermal convection cell.

The convection model described above is based on several simplifying assumptions, such as neglecting adiabatic temperature effects and treating the density as constant except in the buoyancy term (Boussinesq approximation). These assumptions simplify the governing equations and make the problem considerably easier to solve numerically while still capturing the essential physics of thermal convection.

To solve the problem numerically, the governing conservation equations must be solved:

1. conservation of energy (temperature),
2. conservation of momentum, and
3. conservation of mass.

For the two-dimensional case considered here, with heating from below and without viscous dissipation, the equations are given by:

*Conservation of temperature*

$\begin{equation}
\rho c_p \left(\frac{\partial T}{\partial t} + v_j \frac{\partial{T}}{\partial{x_j}}\right) = -\frac{\partial q_i}{\partial x_i},
\end{equation}$

where $\rho$ is the density in [kg/m³], $c_p$ is the specific heat capacity in [J/kg/K], $T$ is the temperature in [K], $t$ is time in [s], $\frac{\partial}{\partial{t}}$ denotes the time derivative, $v_j$ is the velocity component in the $j$-th spatial direction in [m/s], $q_i$ is the heat flux in the $i$-th direction in [W/m²], and $\frac{\partial}{\partial{x_i}}$ denotes the derivative with respect to the $i$-th spatial coordinate. In this general form of the energy equation, all material parameters may vary in space and time, while temperature changes caused by pressure effects are neglected, that is, no adiabatic temperature changes are considered.

*Conservation of momentum*

$\begin{equation}
\rho \left(\frac{\partial{v_{i}}}{\partial{t}} + v_{j}\frac{\partial{v_{i}}}{\partial{x_{j}}}\right) = -\frac{\partial{P}}{\partial{x_{i}}} + \frac{\partial{\tau_{ij}}}{\partial{x_j}} + \rho g_{i},
\end{equation}$

where $P$ is the total pressure $\left(P = P_{\mathrm{dynamic}} + P_{\mathrm{hydrostatic}}\right)$ in [Pa], and $\tau_{ij}$ is the deviatoric stress tensor in [Pa].

The deviatoric stress tensor is given by:

$\begin{equation}
\tau_{ij} = 2 \eta \cdot \dot{\varepsilon}_{ij},
\end{equation}$

where $\eta$ is the dynamic viscosity in [Pa·s], and $\dot{\varepsilon}_{ij}$ is the strain-rate tensor in [1/s], defined as:

$\begin{equation}
\dot{\varepsilon}_{ij} = \frac{1}{2} \left(\frac{\partial{v_i}}{\partial{x_j}} + \frac{\partial{v_j}}{\partial{x_i}}\right).
\end{equation}$

In this general form of the momentum equation, the material parameters may also vary, while the material is assumed to be incompressible.

Under the incompressibility assumption, conservation of mass is expressed by the continuity equation:

*Continuity equation*

$\begin{equation}
\frac{\partial{v_i}}{\partial{x_i}} = 0.
\end{equation}$

### Boussinesq Approximation

Because of the very high viscosity and the correspondingly small Reynolds number, the inertial terms are neglected. The momentum equation therefore reduces to the stationary Stokes equation:

$\begin{equation}
0 = -\frac{\partial{P}}{\partial{x_{i}}} + \frac{\partial{\tau_{ij}}}{\partial{x_j}} + \rho g_{i}.
\end{equation}$

A commonly applied approximation in models of thermal mantle convection is the so-called Boussinesq approximation. In this approximation, adiabatic temperature effects are neglected, and the material parameters $c_p$, $\rho_0$, and $\alpha$ are assumed to be constant. Because the relative density variations in the Earth's mantle are small, they are considered only in the buoyancy term of the Stokes equation. The temperature-dependent density is described by a linearized equation of state:

$\begin{equation}
\rho = \rho_0 \left[1-\alpha \left(T-T_{\textrm{ref}}\right)\right],
\end{equation}$

where $\rho_0$ is constant and represents the reference density at the reference temperature $T_{\textrm{ref}}$.

The total pressure is defined as

$\begin{equation}
P = P_{\textrm{dyn}} + P_{\textrm{hydr}},
\end{equation}$

where the hydrostatic pressure gradient is given by

$\begin{equation}
\frac{\partial{P_{\textrm{hydr}}}}{\partial{y}} = \rho_0 g.
\end{equation}$

After decomposing the pressure into hydrostatic and dynamic components, the constant reference-density contribution is balanced by the hydrostatic pressure gradient. Therefore, only the temperature-dependent buoyancy contribution remains relevant to the flow:

$\begin{equation}
0 = -\frac{\partial{P_{\textrm{hydr}}}}{\partial{y}} -\frac{\partial{P_{\textrm{dyn}}}}{\partial{y}} + \frac{\partial{\tau_{yj}}}{\partial{x_j}}+\rho_0 g - \rho' g,
\end{equation}$

where the temperature-dependent density anomaly is defined as

$\begin{equation}
\rho' = -\rho_0 \alpha\left(T-T_{\textrm{ref}}\right).
\end{equation}$

Because $T_{\textrm{ref}}$ is spatially constant, the term $\rho_0 g \alpha T_{\textrm{ref}}$ contributes only to a constant hydrostatic pressure gradient. This contribution can be incorporated into the reference pressure, so that only the temperature-dependent buoyancy term remains relevant to the dynamics. Therefore, the numerical model equivalently uses

$\begin{equation}
\rho'=-\rho_0 \alpha T.
\end{equation}$

Here, $T$ denotes the temperature, or temperature anomaly, relevant to buoyancy. The vertical component of the Stokes equation therefore becomes

$\begin{equation}
0 = -\frac{\partial{P_{\textrm{dyn}}}}{\partial{y}} + \frac{\partial{\tau_{yj}}}{\partial{x_j}} - \rho' g.
\end{equation}$

Density and specific heat capacity are also assumed to be constant in the energy equation. These assumptions considerably simplify the conservation equations without altering the essential characteristics of thermal mantle convection. The same reference parameters are used in the following section to scale the equations and convert them into non-dimensional form. This procedure introduces characteristic non-dimensional parameters such as the Rayleigh number.

### *Rayleigh* Number

The behavior of thermal convection can be characterized remarkably well by a single non-dimensional parameter: the *Rayleigh* number. The *Rayleigh* number describes the ratio between the forces driving convection (thermal buoyancy) and the forces resisting convection (viscous resistance and thermal diffusion). When the *Rayleigh* number exceeds a critical value, a thermally stratified fluid becomes unstable, leading to the onset of self-sustaining convective overturn.

The *Rayleigh* number is defined as

$\begin{equation}
Ra = \frac{\rho_0 g \alpha \Delta{T} h^3}{\eta_0 \kappa},
\end{equation}$

where $\rho_0$ is the reference density [kg/m³], $g$ is the gravitational acceleration [m/s²], $\alpha$ is the coefficient of thermal expansion [$K^{-1}$], $\Delta T$ is the temperature difference between the bottom and the top of the layer [K], $h$ is the layer thickness [m], $\eta_0$ is the reference viscosity [Pa·s], and $\kappa = k / \rho_0 / c_p$ is the thermal diffusivity [m²/s]. These quantities also serve as the reference parameters for the convection model.

An important property of the *Rayleigh* number is that it characterizes the convective regime independently of the individual dimensional parameter values. Consequently, laboratory experiments and appropriately scaled numerical models can reproduce the same flow behavior as much larger natural systems, such as convection within the Earth's mantle, provided that they share the same *Rayleigh* number.

As a first example, we consider a moderately vigorous convection with a *Rayleigh* number of $10^4$. The reference parameters are chosen as

$\begin{equation}\begin{split}
H & = 2900\cdot 10^3 \ [m], \\
g & = 9.81 \ [m/s^2], \\
\rho_0 & = 3300.0\ [kg/m^3],\\
k & = 4.125\ [W/m/K],\\
c_p & = 1250.0 \ [J/kg/K],\\
\alpha & = 2.0\times10^{-5} \ [K^{-1}], \\
Q_0 & = 0.0 \ [W/m^3], \\
\eta_0 & = 3.947725485 \times 10^{23} \ [Pa\cdot s], \\
\kappa & = \frac{k}{\rho_0 c_p} = 10^{-6} \ [m^2/s],\\
\Delta T & = 2500.0 \ [K]. \\
\end{split}\end{equation}$

These reference parameters approximately correspond to average values for the Earth's mantle. The reference viscosity is one of the least well constrained material properties and has a particularly strong influence on the *Rayleigh* number because it appears in the denominator of its definition.

The script is designed to provide two alternative ways of defining the convective regime. Either the desired *Rayleigh* number can be specified directly, in which case the reference viscosity is automatically adjusted to obtain the prescribed value, or the individual physical parameters can be modified manually, allowing the resulting *Rayleigh* number to be determined from the chosen material properties.

The governing equations can now be solved within the time loop. The following steps must be considered:

1. Update the model time.
2. Initialize the velocity and pressure fields to zero for `it == 1`.
3. Solve the momentum equation, either directly or using defect correction.
4. Update the velocity and pressure fields.
5. Interpolate the velocity components to the cell centers.
6. Calculate the diagnostic and statistical quantities.
7. Determine and, if necessary, adjust the maximum stable time-step size.
8. Visualize the velocity and temperature fields.
9. Solve the advection equation.
10. Solve the diffusion equation.
11. Check whether the model has reached a statistically steady state.

```julia
# Time Loop ========================================================= #
for it = 1:T.itmax
    if it>1
        Time[it]  =   Time[it-1] + T.Δ[1]
    end
    @printf("Time step: #%04d, Time [Myr]: %04e\n ",it,
                    Time[it]/(60*60*24*365.25)/1.0e6)
    # ======= Momentum Equation(ME) =======
    if it == 1
        D.vx    .=  0.0
        D.vy    .=  0.0 
        D.Pt    .=  0.0
    end
    if FD.Method.Mom==:direct
        # Update K ---
        KM      =   ?
        # Update RHS ---
        # rhs term defined by the Boussinesq approximation
        rhsM    =   ?
        # Solve System of Equations ---
        χ       =   ?
        # Update Unknown Variables ---
        D.vx[:,2:end-1]     .=  ?
        D.vy[2:end-1,:]     .=  ?
        D.Pt                .=  ?
    elseif FD.Method.Mom==:dc
        @. D.ρ  =   ?
        for iter = 1:niterM
            # Initial Residual -------------------------------------------------- #
            Residuals2Dc!(?)
            rhsM[Num.Vx]    =   ?
            rhsM[Num.Vy]    =   ?
            rhsM[Num.Pt]    =   ?
            @printf("||R_M|| = %1.4e\n", norm(rhsM)/sqrt(length(rhsM)))
            norm(rhsM)/sqrt(length(rhsM)) < ϵM ? break : nothing
            # ------------------------------------------------------------------- #
            # Assemble Coefficients ============================================= #
            KM      =   ?
            # ------------------------------------------------------------------- #
            # Solution of the linear system ===================================== #
            χ      .=   ?
            # ------------------------------------------------------------------- #
            # Update Unknown Variables ========================================== #
            D.vx[:,2:end-1]     .+=  ?
            D.vy[2:end-1,:]     .+=  ?
            D.Pt                .+=  ?
        end
        @. D.ρ  =   P.ρ₀
    end
    # ======
    # Get the velocity on the centroids ------
    for i = 1:NC.x
        for j = 1:NC.y
            D.vxc[i,j]  = ?
            D.vyc[i,j]  = ?
        end
    end
    @. D.vc        = ?
    # ---
    # Heat flow at the surface ====================================== #
    # Get temperature at the vertices 
    @. Tv1  =   (D.T_ex[1:end-1,end] + D.T_ex[2:end,end] + 
                    D.T_ex[1:end-1,end-1] + D.T_ex[2:end,end-1])/4
    @. Tv2  =   (D.T_ex[1:end-1,end-1] + D.T_ex[2:end,end-1] + 
                    D.T_ex[1:end-1,end-2] + D.T_ex[2:end,end-2])/4
    # Calculate temperature gradient --- 
    @. dTdy =   -(Tv1 - Tv2)/Δ.y
    # Calculate Nusselt number ---
    Nu[it]     =   0.0
    # Trapezoidal integration -
    for i = 1:NV.x
        if i == 1 || i == NV.x
            afac = 1
        else
            afac = 2
        end
        Nu[it]     += afac * dTdy[i]
    end
    Nu[it]     *=   Δ.x/2
    # Scale Nusselnumber ---
    Nu[it] *=  (M.ymax - M.ymin) / ((M.xmax - M.xmin) * P.ΔT)
    # Mean Temperature ---
    meanT[it]   =   mean(D.T)
    # Root Mean Square Velocity ---
    meanV[it]   =   sqrt(mean(D.vxc.^2 .+ D.vyc.^2))
    # --------------------------------------------------------------- #
    # Calculate time stepping ======================================= #
    T.Δc[1]     =   ?
    T.Δd[1]     =   ?
    T.Δ[1]      =   ?

    if Time[it] + T.Δ[1] >= T.tmax[1]
        T.Δ[1]      =   T.tmax[1] - Time[it]
        final_step  =   1
    end
    # Plot ========================================================== #
    if mod(it,100) == 0 || final_step == 1 || it == 1
        p = heatmap(x.c./1e3,y.c./1e3,D.T',
                xlabel="x[km]",ylabel="y[km]",colorbar=true,
                title="Temperature",color=cgrad(:lajolla),
                aspect_ratio=:equal,xlims=(M.xmin/1e3, M.xmax/1e3),
                ylims=(M.ymin/1e3, M.ymax/1e3),
                layout=(2,1),subplot=1)
        heatmap!(p,x.c./1e3,y.c./1e3,D.vc',color=:imola,
            xlabel="x[km]",ylabel="y[km]",colorbar=true,
            title="Velocity",aspect_ratio=:equal,
            xlims=(M.xmin/1e3, M.xmax/1e3), 
            ylims=(M.ymin/1e3, M.ymax/1e3),
            layout=(2,1),subplot=2)
        contour!(p,x.c./1e3,y.c./1e3,D.T',lw=1,
                    color="white",cbar=true,alpha=0.5,
                layout=(2,1),subplot=1)     
        quiver!(p,x.c2d[1:Pl.qinc:end,1:Pl.qinc:end]./1e3,
            y.c2d[1:Pl.qinc:end,1:Pl.qinc:end]./1e3,
            quiver=(D.vxc[1:Pl.qinc:end,1:Pl.qinc:end].*Pl.qsc,
                    D.vyc[1:Pl.qinc:end,1:Pl.qinc:end].*Pl.qsc),
            la=0.5,color="black",
            layout=(2,1),subplot=2)
        if save_fig == 1
            Plots.frame(anim)
        elseif save_fig == 0
            display(p)
        end
    end
    # --------------------------------------------------------------- #
    # Advection ===================================================== #
    if FD.Method.Adv==:upwind
        ?
    elseif FD.Method.Adv==:semilag
        if it == 1
            @. D.vxco  =   D.vxc
            @. D.vyco  =   D.vyc
        end
        ?
        @. D.vxco  =   D.vxc
        @. D.vyco  =   D.vyc
    elseif FD.Method.Adv==:tracers
        # Advect tracers ---
        @printf("Running on %d thread(s)\n", nthreads())  
        @. ΔT_grid     =   D.T_ex - D.Told_ex
        @threads for k = 1:nmark
            local ΔTm       =   FromCtoM(ΔT_grid, k, Ma, x, y, Δ, NC)
            Ma.T[k]     += ΔTm
        end
        AdvectTracer2D(Ma,nmark,D,x,y,T.Δ[1],Δ,NC,rkw,rkv)
        CountMPC(Ma,nmark,MPC,M,x,y,Δ,NC,NV)
        # Interpolate phase from tracers to grid ---
        Markers2Cells(Ma,nmark,MAVG.PC_th,D.T_ex,MAVG.wte_th,D.wte,x,y,Δ,Aparam,0)           
        D.T         .=  D.T_ex[2:end-1,2:end-1]
        D.Told_ex   .=  D.T_ex
    end
    # --------------------------------------------------------------- #
    # Diffusion ===================================================== #
    if FD.Method.Diff==:explicit
        ?
    elseif FD.Method.Diff==:implicit
        ?
    elseif FD.Method.Diff==:CNA
        ?
    elseif FD.Method.Diff==:ADI
        ?
    elseif FD.Method.Diff==:dc
        D.T0    .=  D.T
        for iter = 1:niter
            # Evaluate residual
            ComputeResiduals2Dc!( ? )
            @printf("||R_T|| = %1.4e\n", norm(R)/sqrt(length(R)))
            norm(R)/sqrt(length(R)) < ϵT ? break : nothing
            # Assemble linear system
            K  = ?
            # Solve for temperature correction: Cholesky factorisation
            Kc = cholesky(K.cscmatrix)
            # Solve for temperature correction: Back substitutions
            δT .= ?
            # Update temperature
            @. D.T += ?
        end        
        # Update extended field for advection scheme --- 
        D.T_ex[2:end-1,2:end-1]     .=  D.T
    end
    # --------------------------------------------------------------- #
    # Check break =================================================== #
    # If the maximum time is reached or if the models reaches steady, 
    # state the time loop is stoped! 
    if Time[it]/1e6/T.year > 1000     # [ Ma ]
        epsC    =   1e-4; 
        ind     =   findfirst(Time./1e6/T.year .> 
                        (Time[it]/1e6/T.year - 1000))
        epsV    =   std(meanV[ind:it])
        if count == 0
            epsV0   = max(epsV, eps(Float64))
            count   += 1
        end
        epsV        /= epsV0
        epsV_history[it] = epsV
        find    =   it
        @printf("ε_Vr = %g, ε_Cr = %g \n",epsV,epsC)
        if Time[it] >= T.tmax[1]
            @printf("Maximum time reached!\n")
            find    =   it
            break
        elseif (epsV <= epsC)
            @printf("Convection reaches steady state!\n")
            find    =   it
            break
        end
    end
    # --------------------------------------------------------------- #
    @printf("\n")
end
```
