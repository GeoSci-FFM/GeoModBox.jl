# Thermal Convection – Non-Dimensional Form

One of the major advantages of numerical and analogue models is that they can be formulated in a dimensionless form through appropriate scaling. This non-dimensionalization allows models of different physical scales to be compared directly. By introducing suitable scaling constants, all physical units can be eliminated from the governing equations. Depending on the assumptions made, a small number of dimensionless parameters remain, which provide a concise description of the physical behavior of the system.

## Problem Statement

We again consider thermal convection, as introduced in the [previous exercise](./11_2D_Thermal_Convection.ipynb).

Under the same assumptions (e.g., the *Boussinesq* approximation, incompressibility, and basal heating), the governing equations for the two-dimensional case are given by:

**Energy equation:**

$\begin{equation}
\rho_0 c_p \left(\frac{\partial T}{\partial t} + v_j \frac{\partial T}{\partial x_j}\right) = -\frac{\partial q_i}{\partial x_i},
\end{equation}$

where

- $\rho_0$ is the reference density [kg/m³],
- $c_p$ is the specific heat capacity [J/kg/K],
- $T$ is the temperature [K],
- $t$ is time [s],
- $v_j$ is the velocity component in the $j$-th spatial direction [m/s], and
- $q_i$ is the heat flux in the $i$-th direction [W/m²].

**Momentum equation:**

$\begin{equation}
0 = -\frac{\partial P_{\mathrm{dyn}}}{\partial x_i} +
\frac{\partial \tau_{ij}}{\partial x_j} - \rho_0 g_i \alpha T,
\end{equation}$

where

- $P_{\mathrm{dyn}}$ is the dynamic pressure [Pa],
- $\tau_{ij}$ is the Cauchy stress tensor [Pa],
- $g_i$ is the component of the gravitational acceleration vector [m/s²], and
- $\alpha$ is the coefficient of thermal expansion [1/K].

**Continuity equation:**

$\begin{equation}
\frac{\partial v_i}{\partial x_i}=0.
\end{equation}$

## Scaling

The purpose of non-dimensionalization is to reduce the number of independent physical parameters. Instead of many dimensional material properties, the behavior of the system is described by only a few dimensionless quantities, most notably the **Rayleigh number**.

### Scaling Constants

There are several possible ways to non-dimensionalize the governing equations. The most important step is the selection of suitable scaling constants. Here we define

$$
\begin{aligned}
h_{\mathrm{sc}} &= h, \\
t_{\mathrm{sc}} &= \frac{h^2}{\kappa}, \\
v_{\mathrm{sc}} &= \frac{\kappa}{h}, \\
\tau_{\mathrm{sc}} &= \frac{\eta_0\kappa}{h^2}, \\
T_{\mathrm{sc}} &= \Delta T, \\
Q_{\mathrm{sc}} &= \frac{\Delta T\,\kappa\,\rho_0\,c_p}{h^2},
\end{aligned}
$$

where

- $h$ is the model height [m],
- $\kappa = \dfrac{k}{\rho_0 c_p}$ is the thermal diffusivity [m²/s],
- $\eta_0$ is the reference viscosity [Pa·s], and
- $\Delta T$ is the temperature difference between the bottom and top boundaries [K].

### Scaling Relations

Using the scaling constants defined above, the dimensional variables are transformed into dimensionless quantities (primed variables denote dimensionless quantities):

$$
\begin{aligned}
h &= h' h_{\mathrm{sc}}, \\
t &= t' t_{\mathrm{sc}}, \\
v &= v' v_{\mathrm{sc}}, \\
\tau &= \tau' \tau_{\mathrm{sc}}, \\
T &= T' T_{\mathrm{sc}}, \\
Q &= Q' Q_{\mathrm{sc}}.
\end{aligned}
$$

Applying these scaling relations to Equations (1)–(3) eliminates most physical constants. The resulting equations are dimensionless and depend only on a small number of **dimensionless control parameters**.

> **Note:** The following simplification is valid only under the assumptions adopted in this exercise.

### Rayleigh Number

The most important remaining parameter is the **Rayleigh number**, defined as

$\begin{equation}
Ra=\frac{\rho_0 g \alpha \Delta T h^3}
{\eta_0\kappa}.
\end{equation}$

The Rayleigh number describes the ratio between the forces driving convection (thermal buoyancy) and those resisting convection (viscosity and thermal diffusion). When the Rayleigh number exceeds a critical value, a thermally stratified layer becomes unstable and convection begins.

Regardless of the absolute values of the individual physical parameters, a given Rayleigh number always represents the same convection regime. Consequently, scaled numerical or laboratory models can be used to reproduce large-scale processes such as thermal convection in the Earth's mantle.

The script is designed so that either the Rayleigh number can be specified directly (in which case the reference viscosity $\eta_0$ is adjusted automatically), or the reference parameters can be modified to obtain a desired Rayleigh number.

For a Rayleigh number of $10^4$, we use the following reference values:

$$
\begin{aligned}
g &= 9.81~\mathrm{m/s^2},\\
\rho_0 &= 3300.0~\mathrm{kg/m^3},\\
k &= 4.125~\mathrm{W/m/K},\\
c_p &= 1250.0~\mathrm{J/kg/K},\\
\alpha &= 2.0\times10^{-5}~\mathrm{K^{-1}},\\
Q_0 &= 0.0~\mathrm{W/m^3},\\
\eta_0 &= 3.947725485\times10^{23}~\mathrm{Pa\cdot s},\\
\kappa &= 10^{-6}~\mathrm{m^2/s},\\
\Delta T &= 2500.0~\mathrm{K}.
\end{aligned}
$$

## Dimensionless Governing Equations

The resulting dimensionless equations are

### Momentum equation

$x$-component:

$\begin{equation}
-\frac{\partial P'}{\partial x'} +
\frac{\partial \tau'_{xj}}{\partial x'_j} = 0,
\end{equation}$

or, expressed in terms of the velocity components,

$\begin{equation}
-\frac{\partial P'}{\partial x'} +
2\frac{\partial}{\partial x'} \eta'
\frac{\partial v_x'}{\partial x'} +
\frac{\partial}{\partial y'} \eta'
\left(\frac{\partial v_x'}{\partial y'} +
\frac{\partial v_y'}{\partial x'}\right) = 0.
\end{equation}$

$y$-component:

$\begin{equation}
-\frac{\partial P'}{\partial y'} +
\frac{\partial \tau'_{yj}}{\partial x'_j} = Ra\,T',
\end{equation}$

or

$\begin{equation}
-\frac{\partial P'}{\partial y'} +
2\frac{\partial}{\partial y'} \eta'
\frac{\partial v_y'}{\partial y'} +
\frac{\partial}{\partial x'} \eta'
\left(\frac{\partial v_y'}{\partial x'} +
\frac{\partial v_x'}{\partial y'}
\right) = Ra\,T'.
\end{equation}$

### Energy equation

$\begin{equation}
\left(
\frac{\partial T'}{\partial t'} + v_j'
\frac{\partial T'}{\partial x_j'}\right) =
\frac{\partial^2 T'}{\partial x_i'^2}.
\end{equation}$

### Continuity equation

$\begin{equation}
\frac{\partial v_x'}{\partial x'} +
\frac{\partial v_y'}{\partial y'} = 0.
\end{equation}$

Compared to the dimensional formulation, the numerical solution procedure remains unchanged. The primary difference is that the buoyancy term in the momentum equation is replaced by the product of the **Rayleigh number** and the **dimensionless temperature**. Consequently, the Rayleigh number becomes the principal parameter controlling the dynamics of the convection model.

### Technical Note on the Model Implementation

Unlike the previous exercises, this example makes use of **mutable structures** in `Julia`.

These structures are required because the model parameters must be **modified during the scaling procedure**, which is not possible with the **named tuples** used in the previous exercises, as tuples in Julia are **immutable**, particularly for scalar values.

The mutable structures must be defined separately, for example in a module. Their definitions explicitly specify

- the individual **fields** (e.g., `xmax`, `ymin`), and
- the corresponding **data types** (scalar, vector, matrix, etc.).

Default values can also be assigned to each field. Consequently, it is usually not necessary to specify every parameter explicitly when creating a new structure, since the default values are often sufficient.

The definitions of these structures can be found here:

📁 [`Structures.jl`](../../src/Structures.jl) or in the [documentation](https://geosci-ffm.github.io/GeoModBox.jl/dev/man/Ini/).

To use these structures, simply load the `GeoModBox` module. For example,

```julia
M = Geometry(
    xmax = 8700e3,     # [m]
    ymin = -2900e3     # [m]
)
```

In addition to introducing mutable structures, the numerical solution procedures have been optimized compared to the previous exercise. Since the governing equations are linear and both the viscosity and the thermal properties are assumed to be constant, the assembly and factorization of the coefficient matrices only need to be performed once rather than during every iteration. For the momentum equation, the coefficient matrix can even be assembled before entering the time loop because the viscosity remains constant throughout the simulation.

Furthermore, the amount of information printed to the screen has been reduced (`verbose_step`), which also noticeably improves the overall performance of the time-stepping loop.

This means that we must compute the time step length according to:  

- the **diffusion stability criterion**, and  
- the **Courant criterion**.  

The maximum time step is then given by the **minimum** of the two.  

For the maximum time and the maximum number of iterations, we initially choose values *ad hoc*.  
In practice, the simulation should stop once the **average velocity no longer changes statistically**, indicating that the system has reached a steady state.

Now we can solve the equations inside a time loop. The following steps must be carried out:

1. Initialize the unknown vector and right-hand side for the momentum equation  
2. Compute the current time  
3. Zero the velocity and pressure fields  
4. Solve the momentum equations  
5. Update the velocity and pressure fields  
6. Compute velocities at the *centroids*  
7. Adjust the maximum time step  
8. Visualize velocity, temperature, and density  
9. Solve the advection equation  
10. Solve the diffusion equation  
11. Compute diagnostic measures (e.g., **Nusselt** number, $V_{RMS}$, etc.)  

````julia
# Time loop ========================================================= #
for it = 1:T.itmax
    R0      =   0
    verbose_step    =   mod(it, 250) == 0 || it == 1 || final_step == 1
    if it>1
        Time[it]  =   Time[it-1] + T.Δ
    end
    verbose_step && @printf("Time step: #%04d, Time [non-dim]: %04e\n ",it,
                    Time[it])

    # Momentum equation --------------------------------------------- #
    verbose_step && @printf("---Momentum Calculation ---\n")
    if it == 1
        D.vx    .=  0.0
        D.vy    .=  0.0 
        D.Pt    .=  0.0
    end
    if FD.Method.Mom==:direct
        # Update RHS ---
        # Right-hand side defined by the Boussinesq approximation
        rhsM    =   updaterhsc( ? )
        # Solve the linear system of equations ---
        χ       =   ?
        # Update unknown variables ---
        D.vx[:,2:end-1]     .=  ?
        D.vy[2:end-1,:]     .=  ?
        D.Pt                .=  ?
    elseif FD.Method.Mom==:dc
        @. D.ρ  =   ?
        # Initial residual ---
        for iter = 1:niterM
            Residuals2Dc!( ? )
            rhsM[Num.Vx]    .=   ?
            rhsM[Num.Vy]    .=   ?
            rhsM[Num.Pt]    .=   ?
            RM          =   norm(rhsM)/length(rhsM)
            if iter == 1
                R0 = max(RM, eps())
            end
            RMrel       =   RM/R0
            if verbose_step
                @printf("   MCE %2d: ||R|| = %1.4e, ||R||/||R₀|| = %1.4e\n",iter,RM,RMrel)
            end
            (RM < atolM || RM/R0 < rtolM) && break
            # Solve the linear system of equations ---
            χ      .=   - ( ? )
            # Update unknown variables ---
            D.vx[:,2:end-1]     .+=  ?
            D.vy[2:end-1,:]     .+=  ?
            D.Pt                .+=  ?
        end
    end

    # ======
    # Calculate velocity at the cell centers ------------------------ #
    for i = 1:NC.x
        for j = 1:NC.y
            D.vxc[i,j]  = (D.vx[i,j+1] + D.vx[i+1,j+1])/2
            D.vyc[i,j]  = (D.vy[i+1,j] + D.vy[i+1,j+1])/2
        end
    end
    @. D.vc        = sqrt(D.vxc^2 + D.vyc^2)

    # ---
    # Heat flow at the surface ====================================== #
    # Get temperature at the vertices 
    @. Tv1  =   (D.T_ex[1:end-1,end] + D.T_ex[2:end,end] + 
                    D.T_ex[1:end-1,end-1] + D.T_ex[2:end,end-1])/4
    @. Tv2  =   (D.T_ex[1:end-1,end-1] + D.T_ex[2:end,end-1] + 
                    D.T_ex[1:end-1,end-2] + D.T_ex[2:end,end-2])/4
    # Calculate temperature gradient --- 
    @. dTdy =   -(Tv1 - Tv2)/Δ.y
    # Calculate Nusselt number ---
    Nu[it]     =   0.0
    # Trapezoidal integration -
    for i = 1:NV.x
        if i == 1 || i == NV.x
            afac = 1
        else
            afac = 2
        end
        Nu[it]     += afac * dTdy[i]
    end
    Nu[it]     *=   Δ.x/2
    Nu[it]     /= (M.xmax - M.xmin)

    # Mean temperature ---
    meanT[it]   =   mean(D.T)

    # Root mean square velocity ---
    meanV[it] = sqrt(mean(D.vxc.^2 .+ D.vyc.^2))

    # --------------------------------------------------------------- #
    # Calculate the time-step size ================================== #
    T.Δc        =   ?
    T.Δd        =   ?
    T.Δ         =   ?
    if Time[it] + T.Δ >= T.tmax
        T.Δ         =   T.tmax - Time[it]
        final_step  =   1
    end

    # Plot ========================================================== #
    if verbose_step
        p = plot(layout=(2,1),size=(1200,800),dpi = 300,
            guidefontsize = 22, tickfontsize = 22,
            titlefontsize = 22)
        heatmap!(p,x.c,y.c,D.T',
                xlabel="",ylabel=L"y",colorbar=true,
                framestyle =:box,
                xformatter = _ -> "",
                left_margin = 10mm,
                right_margin = 10mm, 
                title=L"Temperature",color=cgrad(:lajolla),
                aspect_ratio=:equal,xlims=(M.xmin, M.xmax),
                ylims=(M.ymin, M.ymax),clims=(0,1),
                subplot=1)
        heatmap!(p,x.c,y.c,D.vc',color=:imola,
            xlabel=L"x",ylabel=L"y",colorbar=true,
            framestyle =:box,
            title=L"Velocity",aspect_ratio=:equal,
            xlims=(M.xmin, M.xmax), 
            ylims=(M.ymin, M.ymax),
            subplot=2)
        contour!(p,x.c,y.c,D.T',lw=1,
            color="white",cbar=true,alpha=0.5,
            subplot=1)     
        quiver!(p,x.c2d[1:Pl.qinc:end,1:Pl.qinc:end],
            y.c2d[1:Pl.qinc:end,1:Pl.qinc:end],
            quiver=(D.vxc[1:Pl.qinc:end,1:Pl.qinc:end].*Pl.qsc,
                    D.vyc[1:Pl.qinc:end,1:Pl.qinc:end].*Pl.qsc),
            la=0.5,color="black",
            subplot=2)
        if save_fig == 1
            Plots.frame(anim)
        elseif save_fig == 0
            display(p)
        end
    end

    # --------------------------------------------------------------- #
    # Advection ===================================================== #
    if FD.Method.Adv==:upwind
        upwindc2D!( ? )
    elseif FD.Method.Adv==:semilag
        if it == 1
            D.vxco  =   copy(D.vxc)
            D.vyco  =   copy(D.vyc)
        end
        semilagc2D!( ? )
        D.vxco  =   copy(D.vxc)
        D.vyco  =   copy(D.vyc)
    elseif FD.Method.Adv==:tracers
        # Advect tracers ---
        verbose_step && @printf("Running on %d thread(s)\n", nthreads())  
        @. ΔT_grid     =   D.T_ex - D.Told_ex
        @threads for k = 1:nmark
            local ΔTm       =   FromCtoM(ΔT_grid, k, Ma, x, y, Δ, NC)
            Ma.T[k]     += ΔTm
        end
        AdvectTracer2D(Ma,nmark,D,x,y,T.Δ,Δ,NC,rkw,rkv)
        CountMPC(Ma,nmark,MPC,M,x,y,Δ,NC,NV)

        # Interpolate temperature from tracers to the grid ---
        Markers2Cells(Ma,nmark,MAVG.PC_th,D.T_ex,MAVG.wte_th,D.wte,x,y,Δ,Aparam,0)           
        D.T         .=  D.T_ex[2:end-1,2:end-1]
        D.Told_ex   .=  D.T_ex
    end

    # --------------------------------------------------------------- #
    # Diffusion ===================================================== #
    verbose_step && @printf("---Energy Calculation---\n")
    if FD.Method.Diff==:explicit
        ?
    elseif FD.Method.Diff==:implicit
        ?
    elseif FD.Method.Diff==:CNA
        ?
    elseif FD.Method.Diff==:ADI
        ?
    elseif FD.Method.Diff==:dc
        D.T0    .=  D.T

        # Assemble the linear system
        K  = AssembleMatrix2Dc( ? )

        # Factorize the coefficient matrix using Cholesky factorization
        Kc = cholesky(K.cscmatrix)

        for iter = 1:niterT
            # Evaluate the residual
            ComputeResiduals2Dc!( ? )
            verbose_step && @printf("||R_T|| = %1.4e\n", norm(R)/sqrt(length(R)))
            norm(R)/sqrt(length(R)) < ϵT ? break : nothing

            # Solve for the temperature correction using back substitution
            δT .= -( ? )

            # Update temperature
            @. D.T += ?
        end        

        # Update the extended temperature field for the advection scheme ---
        D.T_ex[2:end-1,2:end-1]     .=  D.T
    end

    # --------------------------------------------------------------- #
    # Check stopping criteria ======================================= #
    # Stop the time loop when the maximum time is reached or when
    # the model reaches a statistically steady state.
    if Time[it] > 0.0038
        epsC    =   0.0001
        ind = searchsortedfirst(@view(Time[1:it]), Time[it] - 0.0038)
        epsV    =   std(@view meanV[ind:it])
        if count == 0
            epsV0   = max(epsV, eps(Float64))
            count   += 1
        end
        epsV        /= epsV0
        epsV_history[it] = epsV
        find    =   it
        verbose_step && @printf("ε_Vr = %g, ε_Cr = %g \n",epsV,epsC)
        if Time[it] >= T.tmax
            @printf("Maximum time reached!\n")
            find    =   it
            break
        elseif (epsV <= epsC)
            @printf("Convection reaches steady state at %i!\n",it)
            find    =   it
            break
        end
    end

    # --------------------------------------------------------------- #
    verbose_step && @printf("\n")
end
````
