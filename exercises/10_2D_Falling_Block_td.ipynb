# Stokes-Gleichung (2D)

## Zeitabhängiger fallender Block bei konstanter Viskosität

In dieser Übung erweitern wir das zuvor betrachtete instantane Problem des
fallenden Blocks um die zeitliche Entwicklung der Dichteverteilung.

Dazu koppeln wir die Lösung der Impuls- und Massenerhaltung mit einer
Advektionsmethode. In jedem Zeitschritt wird zunächst das momentane
Geschwindigkeits- und Druckfeld für die aktuelle Dichteverteilung berechnet.
Anschließend wird die Dichteverteilung mit diesem Geschwindigkeitsfeld
transportiert.

Für die Advektion stehen mehrere Verfahren zur Verfügung:

- Upwind,
- Staggered Leapfrog,
- Semi-Lagrange,
- passive Tracer.

Bei den gitterbasierten Verfahren wird die Dichte direkt advektiert. Bei der
Tracermethode transportieren die Marker hingegen die Materialphase, aus der
anschließend die Dichte auf dem numerischen Gitter rekonstruiert wird.

Da die Viskosität räumlich und zeitlich konstant bleibt, ändert sich die
Koeffizientenmatrix des Stokes-Systems nicht. Sie kann daher einmal vor der
Zeitschleife assembliert werden. Innerhalb der Zeitschleife
muss lediglich der von der aktuellen Dichteverteilung abhängige Vektor der
rechten Seite aktualisiert werden.

### Das Problem

Anders als im vollständigen Benchmark für dieses Problem, konzentrieren wir uns hier nur auf den **isoviskosen** Fall (im Benchmark betrachtet man die Sinkgeschwindigkeit und Deformation des Blocks in Abhängigkeit des Viskositätskontrastes; [siehe hier](../examples/StokesEquation/2D/)). 

Dabei nehmen wir einen quadratischen Körper mit einer gewissen Breite (**W**), Höhe (**T**), und Dichte ($\rho_b$) an, der sich in einem bestimmten viskosen Medium ($\eta_m$) mit einer gewissen Dichte ($\rho_m$) befindet. Die zu modellierende Umgebung ist ebenfalls quadratisch mit einer gewissen Länge ($L=500\ \textrm{km}$) und Höhe ($H=500\ \textrm{km}$) und wir nehmen überall *free slip* Geschwindigkeitsrandbedingungen an.

Für die tracerbasierte Advektion müssen die passiven Tracer zunächst initialisiert werden.. Die Funktion `IniTracer2D()` initialisiert die Verteilung der Tracer in der gegebenen Modelldomäne (mit, wenn nötig, beliebiger Störung bei der Position) und, falls vorher in `Ini` definiert, wird den Tracern die jeweilige Phase zugewiesen. 

Die Tracer werden in der Zeitschleife advektiert und transportieren die Phase. Mithilfe der auf den Tracern gespeicherten Phase wird anschließend eine Materialeigenschaft, hier die Dichte ρ, mit der Funktion `Markers2Cells()` auf das numerische Gitter interpoliert.

Falls eine andere Advektionsmethode gewählt wird, wird die Dichte direkt auf dem Gitter initialisiert. Beachte, die Advektionsroutinen benötigen die Information auf dem erweiterten Gitter. Dabei weist `IniPhase!()` jedem Zellzentrum die entsprechende Materialphase zu. Anschließend wird jeder Phase der zugehörige Dichtewert zugewiesen. Für das Staggered-Leapfrog-Verfahren wird zusätzlich die Dichteverteilung des vorherigen Zeitschritts in `D.ρ_exo` gespeichert.

```julia
# Tracer Advection ================================================== #
if FD.Method.Adv==:tracers 
    # Tracer Initialization ---
    nmx,nmy     =   5,5
    noise       =   0
    nmark       =   nmx*nmy*NC.x*NC.y
    Aparam      =   :phase
    MPC         =   (
            c       =   zeros(Float64,(NC.x,NC.y)),
            v       =   zeros(Float64,(NV.x,NV.y)),
            th      =   zeros(Float64,(nthreads(),NC.x,NC.y)),
            thv     =   zeros(Float64,(nthreads(),NV.x,NV.y)),
    )
    MAVG        = (
            PC_th   =   [similar(D.wte) for _ = 1:nthreads()],  # per thread
            PV_th   =   [similar(D.wtv) for _ = 1:nthreads()],   # per thread
            wte_th  =   [similar(D.wte) for _ = 1:nthreads()],  # per thread
            wtv_th  =   [similar(D.wtv) for _ = 1:nthreads()],  # per thread
    )
    Ma      =   IniTracer2D(Aparam,nmx,nmy,Δ,M,NC,noise,Ini.p,phase)
    # RK4 weights ---
    rkw     =   1.0/6.0*[1.0 2.0 2.0 1.0]   # for averaging
    rkv     =   1.0/2.0*[1.0 1.0 2.0 2.0]   # for time stepping
    # Count marker per cell ---
    CountMPC( ? )
    # Interpolate from markers to cell ---
    Markers2Cells( ? )
    D.ρ     .=  D.ρ_ex[2:end-1,2:end-1]  
else
    # --------------------------------------------------------------- #
    # Initial Condition ============================================= #
    # Phase ---
    # If tracers are used, phases need to be defined on the tracers 
    # directly and are not suppose to be interpolated from the centroids! 
    IniPhase!( ? )
    for i in eachindex(phase)
        D.ρ[D.p.==phase[i]] .= ρ[i]
    end
    D.ρ_ex[2:end-1,2:end-1]     .=  D.ρ
    D.ρ_ex[1,:]     .=   D.ρ_ex[2,:]
    D.ρ_ex[end,:]   .=   D.ρ_ex[end-1,:]
    D.ρ_ex[:,1]     .=   D.ρ_ex[:,2]
    D.ρ_ex[:,end]   .=   D.ρ_ex[:,end-1]
    D.ρ_exo         .=   D.ρ_ex
end
# ------------------------------------------------------------------- #
```

In jedem Zeitschritt werden
